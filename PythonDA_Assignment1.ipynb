Indexing and Slicing:

```python
Marks[0]
```

```python
top_3_marks = Marks.loc[['Rank1', 'Rank2', 'Rank3']]
print(top_3_marks)
```

```python
Marks.iloc[2]
```

Apply a boolean mask to filter the Series and retrieve the ranks where the
marks are greater than 90.

```python
mask = Marks > 90
print(mask)
```
